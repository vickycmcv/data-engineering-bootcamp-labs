```python
nightingale_summary = sc.textFile('/user/training/week5_session4_lab/the_nightingale_summary.txt')
stopwords = sc.textFile('/user/training/week5_session4_lab/stopwords-en.txt')
```

```python
clean_file = nightingale_summary.map(lambda x:x.replace(',', '').replace('.', '').replace('"', '').replace('  ', ''))
clean_file.take(20)
```

```python
flatted_file = clean_file.flatMap(lambda x:x.lower().split(' '))
flatted_file.take(20)
```

```python
file_wo_stopwords = flatted_file.subtract(stopwords)
file_wo_stopwords.take(50)
```

```python
file_wo_stopwords_kv = file_wo_stopwords.map(lambda x:(x,1))
file_wo_stopwords_kv.take(50)
```

```python
count_file_words = file_wo_stopwords_kv.reduceByKey(lambda x,y:x+y)
count_file_words.take(50)
```

```python
sorted_count_file_words = count_file_words.sortBy(lambda x:x[1], False)
sorted_count_file_words.take(20)
```

```python
top5_words_list = sorted_count_file_words.keys().take(5)
top5_words_list
```

```python
top5_words = sc.parallelize(top5_words_list)
top5_words.collect()
```

```python
top5_words.saveAsTextFile('/user/training/week5_session4_lab/top5_words_the_nightingale')
```
